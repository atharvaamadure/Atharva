```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
```

```python
# Assuming df5 is your DataFrame with 27 columns
# Extract features (exclude the target column)
x = df6.drop(columns=['MOST_SEVERE_INJURY'])
y = df6["MOST_SEVERE_INJURY"]
```

```python
# Standardize the features (important for PCA)
scaler = StandardScaler()
x_scaled = scaler.fit_transform(x)
```

```python
# Create a PCA object with desired components (let's say 2 for now)
n_components = 5
pca = PCA(n_components=n_components)
```

```python
# Fit the PCA model to the scaled features
pca.fit(x_scaled)
```

```python
# Transform the features into the new lower-dimensional space
x_pca = pca.transform(x_scaled)
```

```python
# Extract the explained variance ratio
explained_variance = pca.explained_variance_ratio_
```

```python
plt.plot(range(1, n_components + 1), explained_variance, marker='o', linestyle='--', color='b')
plt.xlabel('Principal Component')
plt.ylabel('Explained Variance Ratio')
plt.title('Explained Variance by Principal Components')
plt.grid(True)
plt.show()
```
